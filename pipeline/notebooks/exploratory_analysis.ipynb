# VoxPop MBG — Exploratory Analysis

A short exploration of the raw TikTok comment dataset and the pipeline
outputs. Place the dataset in `../../data/raw/` and run the pipeline before
executing the output cells.

```python
import json
import sys
from collections import Counter
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parents[0] / 'src'))
from voxpop_mbg.clean import clean_text, dedupe_key
from voxpop_mbg.schema import detect_schema
```

```python
raw_path = next((Path.cwd().parents[1] / 'data' / 'raw').glob('*.json'))
records = json.loads(raw_path.read_text(encoding='utf-8'))
schema = detect_schema(records)
print('rows:', len(records))
print('comment column:', schema.comment_column)
print('date column:', schema.date_column)
print('engagement:', schema.engagement_columns)
```

```python
lengths = [len(str(r.get(schema.comment_column, ''))) for r in records]
lengths = [n for n in lengths if n > 0]
print('comments:', len(lengths))
print('mean length:', round(sum(lengths) / len(lengths), 1))
print('max length:', max(lengths))
```

## Pipeline outputs

Load the exported JSON to review sentiment and issue distributions.

```python
data_dir = Path.cwd().parents[1] / 'web' / 'public' / 'data'
sentiment = json.loads((data_dir / 'sentiment_summary.json').read_text(encoding='utf-8'))
issues = json.loads((data_dir / 'issue_summary.json').read_text(encoding='utf-8'))

for row in sentiment['distribution']:
    print(f"{row['label']:24s} {row['count']:6d}  {row['share']:.2%}")
print()
for issue in issues['issues'][:8]:
    print(f"{issue['issue_name']:24s} {issue['count']:6d}  neg={issue['negative_share']:.2%}")
```
